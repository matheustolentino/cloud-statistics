```python
import requests

PATH_RADAR = '../../data/radar/'
PATH_LIDAR = '../../data/lidar/'

url = 'https://cloudnet.fmi.fi/api/files'
payload = {
    'product': 'radar',
    'site': 'granada',
    'dateFrom': '2020-10-01',
    'dateTo': '2020-10-05'
}
metadata = requests.get(url, payload).json()
for row in metadata:
    res = requests.get(row['downloadUrl'])
    with open(row['filename'], 'wb') as f:
        f.write(res.content)
```
