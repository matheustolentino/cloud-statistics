```python
#%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import netCDF4 as nc
from datetime import timedelta
import matplotlib.dates as mdates
from scipy import interpolate, integrate
import scipy.stats as stats
from scipy.special import gamma
from scipy.optimize import curve_fit
import datetime
import matplotlib as mpl
import os
import pandas as pd
```

```python
PATH_SIZDIST = '../data/cloud_sizdist/'
PATH_FIG     = '../figures/'
```

```python
filenames = []
for file in os.listdir(PATH_SIZDIST):
    if file.startswith("cloud"):
        filenames.append(file)
        print(file)
```

```python
def gamma_pdf(x, alf, bet):
    return ( bet**(alf) * x**(alf-1) * np.exp(-bet*x) )/gamma(alf)

def gamma_2pdf(x, alf1, bet1, loc1, scale1, alf2, bet2, loc2, scale2):
    return gamma_pdf(x, alf1, bet1, loc1, scale1) + gamma_pdf(x, alf2, bet2, loc1, scale1)
```

```python
diameter      = np.loadtxt(PATH_SIZDIST+'diameters.csv', delimiter=',')
surface       = 0.24*10**(-2) # mm ^2 -> cm^2
concentration = []
#for file in [filenames[0]]:
for file in filenames:
    data = np.loadtxt(PATH_SIZDIST+file, delimiter=',')
    num  = np.zeros(data[:, 8:].shape)
    for i in range(data.shape[0]):
        app_pas  = data[i, 7]
        volume   = app_pas*100*surface*10 # PAS * Area * time (10s)
                                         # (m s^-1)*100 * cm^2 * s -> cm^3
        num[i,:] = data[i, 8:]/volume
    
    counts_mean = np.mean(data[:, 8:],axis=0)
    counts_mean = counts_mean/integrate.simps(counts_mean, diameter)
    num_mean    = np.mean(num, axis=0)
    num_mean    = num_mean/integrate.simps(num_mean, diameter)
    
    p0beta  = 2.
    p0scale = 1/p0beta
    rmax    = 15 # um
    pcount, covcount = curve_fit(gamma_pdf, diameter, num_mean, 
                                 p0=[12, p0beta, .7, p0scale])
    
    # for i in range()
    pnum, covnum     = curve_fit(gamma_pdf, diameter, counts_mean, 
                                 p0=[12, p0beta, .7, p0scale])
    # pcount = stats.gamma.fit(counts_mean)
    # pnum   = stats.gamma.fit(num_mean)
    
    fig, axs    = plt.subplots(2, sharex=True, figsize=(8, 5))

    f0 = axs[0].plot(diameter, counts_mean, '--ob',markersize=4.)
    f1 = axs[0].plot(diameter, gamma_pdf(diameter,*pcount), '-r',
                     label=r"$\alpha$ = %.1f"%(pcount[0]))
    axs[0].set_ylabel(r'#')
    axs[0].grid()  
    axs[0].legend()
    
    f2 = axs[1].plot(diameter, num_mean, '--ob',markersize=4.)
    f3 = axs[1].plot(diameter, gamma_pdf(diameter,*pnum), '-r',
                    label=r"$\alpha$ = %.1f"%(pnum[0]))
    axs[1].set_ylabel(r'# cm$^{-3}$')
    axs[1].set_xlabel(r'Diameter [$\mu$m]')
    axs[1].set_xlim([2, 30])
    axs[1].grid()
    axs[1].legend()
    plt.suptitle(file)
    plt.show()
    
```
